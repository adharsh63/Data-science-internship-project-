## Resources

This notebook demonstrates a quick Exploratory Data Analysis (EDA) on the classic Titanic dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)
```

### Initial Data Exploration

```python
# Display first five rows
display(df.head())

# Dataset information
df.info()

# Statistical summary
display(df.describe())

# Missing values
display(df.isnull().sum().to_frame(name='Missing Values'))
```

### Data Visualization

```python
# Set theme
sns.set_theme(style="whitegrid")

# Create subplots for survival patterns
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.countplot(data=df, x='survived', ax=axes[0])
axes[0].set_title("Passenger Survival Count")

sns.countplot(data=df, x='sex', hue='survived', ax=axes[1])
axes[1].set_title("Survival by Gender")

plt.tight_layout()
plt.show()
```

```python
# Age distribution and correlations
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(data=df, x='age', kde=True, ax=axes[0])
axes[0].set_title("Age Distribution")

sns.heatmap(df.select_dtypes(include='number').corr(),
            annot=True, cmap='coolwarm', ax=axes[1])
axes[1].set_title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```
