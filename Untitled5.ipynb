```python
import pandas as pd
import numpy as np
from google.colab import files

# --- Create Sample Data to Resolve NameError ---
# Generating sample datetime index
dates = pd.date_range(start="2023-01-01", periods=30, freq="D")

# Creating sample DataFrame 'df'
df = pd.DataFrame({
    "Date": dates,
    "Store": np.random.choice(["Store_A", "Store_B"], size=30),
    "Sales": np.random.randint(1000, 5000, size=30)
})

# Creating sample 'y_test' and 'predictions'
y_test = pd.Series(np.random.randint(1000, 5000, size=10), index=dates[-10:])
predictions = np.random.randint(1000, 5000, size=10)
# ----------------------------------------------

# Save cleaned dataset
df.to_csv("cleaned_retail_sales.csv", index=False)

# Save prediction results
results = pd.DataFrame({
    "Date": y_test.index,
    "Actual_Sales": y_test.values,
    "Predicted_Sales": predictions
})

results.to_csv(
    "retail_sales_predictions.csv",
    index=False
)

# Download both files
files.download("cleaned_retail_sales.csv")
files.download("retail_sales_predictions.csv")

print("Project output files created successfully!")
```
