```python
# Import libraries and load dataset
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)

display(df.head())
```

```python
# Dataset information and summary statistics
print("=== Dataset Info ===")
df.info()

print("\n=== Descriptive Statistics ===")
display(df.describe())
```

```python
# Check for missing values
print("=== Missing Values Count ===")
print(df.isnull().sum())
```

```python
# Survival visualization (Count and Gender Breakdown)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.countplot(data=df, x='survived', ax=axes[0])
axes[0].set_title("Passenger Survival Count")

sns.countplot(data=df, x='sex', hue='survived', ax=axes[1])
axes[1].set_title("Survival by Gender")

plt.tight_layout()
plt.show()
```

```python
# Age distribution and Correlation heatmap
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(data=df, x='age', kde=True, ax=axes[0])
axes[0].set_title("Age Distribution")

sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm', ax=axes[1])
axes[1].set_title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```
